```python
import pandas as pd
import numpy as np

import re
import warnings
warnings.filterwarnings("ignore")
```

```python
# load dataset using pandas
df = pd.read_csv("/content/magic_bricks_data.csv")
```

```python
df
```

```python
# drop the extra index column
df = df.drop('Unnamed: 0', axis = 1)
```

```python
df.info()
```

```python
# checking the location column
text = ""
for i in df['location']:
  text=text + i + "\n"
```

```python
# regex for extracting the bhk from the location column
regex = "[\d]+\s+BHK"
df['BHK'] = df['location'].apply((lambda x: re.findall(regex,x)[0] if re.findall(regex,x) else None))
```

```python
df.head(5)
```

```python
text=""
for i in df[df["BHK"].isna()]["location"]:
  text=text + i + "\n"
```

```python
Price = ""
for i in df['price']:
  Price = Price + i + "\n"
```

```python
# storing the location in city
City = []
for i in df['location']:
  City.append(i)

```

```python
# regex for only taking city in in the city column and deleting other
# and adding new column to the df
regex = "(?:Mumbai|Bangalore|Hyderabad|Ahembdabad|Ahmedabad|Pune|Chennai|Gurgaon|Kolkata|Delhi|Noida)"
df['City'] = df['location'].apply((lambda x: re.findall(regex,x)[0] if re.findall(regex,x) else None))
```

```python
df.head(5)
```

```python
# regex for extracting only the sqft from the area
regex = "Area+\d+"
df['s'] = df['area'].apply((lambda x: re.findall(regex,x)[0] if re.findall(regex,x) else None))
```

```python
df.head(5)
```

```python
# regex for only taking the digits from the sqft from the 's' column
regex = "\d+"
df['sqft'] = df['s'].apply((lambda x: re.findall(regex,x)[0] if re.findall(regex,x) else None))
```

```python
df.head(5)
```

```python
len(df)
```

```python
# regex for extracting only the furnished and unfurnished text from the area column
# and making the new furnished_status column
regex = "(?:furnished|Unfurnished|Furnished|unfurnished)"
df['furnished_status'] = df['area'].apply((lambda x: re.findall(regex,x)[0] if re.findall(regex,x) else None))
```

```python
df.head(5)
```

```python
# droping the extra column after extracting the usefull information
df.drop(columns = ['location','area','s'],inplace = True)
```
