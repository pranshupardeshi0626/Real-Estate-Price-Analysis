###**Price Column Should Be Numeric**


```python
# first replace the special chacracter like '₹' and '.' from epmpty string
# then checked if the value is cr then so add the zeros as per the condition
# then checked if the value is lakh then add the required zero as per the conditon

df["price"] = [i.replace("₹","").replace("."," ") for i in df["price"]]
v = []
for i in df["price"]:
  j = i.split()

  if "Cr" in j:
    if len(j) == 3:
      if len(j[1]) == 1:
          v.append([j[0],j[1],'000000'])
      elif len(j[1]) == 2:
          v.append([j[0],j[1],'00000'])
    elif len(j) == 2:
      v.append([j[0],'0000000'])

  elif "Lac" in j:
    if len(j) == 3:
      if len(j[1]) == 1:
          v.append([j[0],j[1],'0000'])
      elif len(j[1]) == 2:
          v.append([j[0],j[1],'000'])
    elif len(j) == 2:
      v.append([j[0],'00000'])

  else:
    v.append(np.nan)

df["price"]=["".join(i) for i in v]
df["price"]= df['price'].astype(float)
```
