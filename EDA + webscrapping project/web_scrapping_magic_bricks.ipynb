```python
society_name =[]
price_data =[]
location_data=[]
area_data =[]

for i in cities :
    
    
    from curl_cffi import requests 
    
    response = requests.get(i, impersonate = "chrome110")
    response
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/123.0.0.0 Safari/537.36"}
    
    response = requests.get(i,headers=headers)
    html_text = response.text
    soup = BeautifulSoup(html_text)
    
    containers = soup.find_all("div", attrs ={"class" : "mb-srp__list"})

        
    for c in containers :
            
        building_name = c.find("div", attrs = {"class" : "mb-srp__card__society"})
        
        if building_name is None :
            society_name.append(np.nan)
        else:
            society_name.append(building_name.text)
        
            

        price = c.find("div", attrs ={"class" : "mb-srp__card__price--amount"})
        
        if price is None:
            price_data.append(np.nan)
        else:
            price_data.append(price.text)
                
    
        location = c.find("h2", attrs ={"class" : "mb-srp__card--title"})
        
        if location is None :
            location_data.append(np.nan)
        else:
            location_data.append(location.text)
                
    
        carpet_area = c.find("div" ,attrs  = {"class" : "mb-srp__card__summary__list"})
     
        if carpet_area is None :
            area_data.append(np.nan)
        else:
            area_data.append(carpet_area.text)
            
    
       
               

      
```
